```python
import pandas as pd
df = pd.read_json('../data/auto.json')
df = df.set_index('CarNumber')
```

selects из readme: Штрафы больше 2100

```python
fines = df[df['Fines'] > 2100]
```

штрафы и 2 возврата

```python
fines[fines['Refund'] == 2]
```

Модели из списка:

```python
models = ['Focus', 'Corolla']
df[df['Model'].isin(models)]
```

Номера из списка:

```python
target_cars = ['Y7689C197RUS', '92928M178RUS', '7788KT197RUS', 'H115YO163RUS', 'X758HY197RUS']
df[df.index.isin(target_cars)]
```

Медиана штрафов по марке:

```python
df.groupby('Make')['Fines'].median()
```

Медиана по марке и модели:

```python
df.groupby(['Make', 'Model'])['Fines'].median()
```

Количество штрафов по марке и модели, чтобы понять, можно ли доверять медианным значениям:

```python
df.groupby(['Make', 'Model'])['Fines'].count()
```

Минимальные и максимальные штрафы, чтобы лучше понимать разницу:

```python
df.groupby(['Make', 'Model'])['Fines'].agg(['min', 'max'])
```

Стандартное отклонение:

```python
df.groupby(['Make', 'Model'])['Fines'].std()
```

Кто чаще всего нарушал закон:

```python
df.groupby('CarNumber')['Fines'].count().sort_values(ascending=False).head(3)
```

Кто больше всего заплатил:

```python
df.groupby('CarNumber')['Fines'].sum().sort_values(ascending=False).head(1)
#df[df.index == fines_sum.index[0]]
```
