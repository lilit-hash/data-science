```python
import pandas as pd
df = pd.read_csv('../data/auto.csv', index_col='ID')
```

Считаем до удаления дубликатов:

```python
df.count()
```

```python
df = df.drop_duplicates(subset=['CarNumber', 'Make_n_model', 'Fines'], keep='last')
```

Считаем после:

```python
df.count()
```

Пропущенные значения до обработки:

```python
df.isnull().sum()
```

Если в колонке более 500 пропущенных значений, то удаляем:

```python
df = df.dropna(thresh=len(df)-500, axis=1)
```

смотрим новый результат после обработки:

```python
df.isnull().sum()
```

заменяем пропущенные значения возвратов (refund) предыдущими значениями с помощью forward fill

```python
df['Refund'] = df['Refund'].ffill()
ref = df.isnull().sum()
ref
```

Заменяем пропущенные в fines средним значением

```python
mean_fine = df['Fines'].mean()
df['Fines'] = df['Fines'].fillna(mean_fine)
```

```python
df.isnull().sum()
```

делим Make_n_model на Make, Model

```python
df[['Make', 'Model']] = df['Make_n_model'].apply(
    lambda x: pd.Series(str(x).split(' ', 1)) if pd.notna(x) else pd.Series([None, None]))
```

```python
df = df.drop('Make_n_model', axis=1)
```

```python
df.to_json('../data/auto.json', orient='records', indent=2)
```
