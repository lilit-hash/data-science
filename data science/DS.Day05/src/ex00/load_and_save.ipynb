```python
import pandas as pd
```

Читаем файл и пропускаем строки с индексами 2,3, а также две последние строки файла, используем пайтон-движок для skipfooter, первую колонку используем как индекс

```python
df = pd.read_csv('../data/feed-views.log', skiprows=[2,3], skipfooter=2, engine='python', header=None, names=['datetime', 'user'], sep='\t', index_col=0)
```

меняем название индексной колонки

```python
df.index.name = 'date_time'
```

```python
df.to_csv('../data/feed-views-semicolon.log', sep=';', header=False)
```
